```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# 1. Database Connection Setup
username = "postgres"
password = "your_password" # Yahan apna pgAdmin password dalein
host = "localhost"
port = "5432"
database = "vendor_performance_data" # Aapke pichle screenshot wala database

# Engine create karna
engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")
```

```python
# 2. Saari CSV files aur unke Tables ka naam
csv_files = {
    "begin_inventory.csv": "begin_inventory",
    "end_inventory.csv": "end_inventory",
    "purchase_prices.csv": "purchase_prices",
    "purchases.csv": "purchases",
    "vendor_invoice.csv": "vendor_invoice",
    "sales.csv": "sales"  # 1.5 GB ki sabse badi file
}
```

```python
# 3. Data Load karne ka Loop (with Chunking to prevent crash)
for file_name, table_name in csv_files.items():
    print(f"\nLoading '{file_name}' into table '{table_name}'...")
    
    # Badi files ko handle karne ke liye data ko 1 Lakh rows ke chunks mein read karenge
    chunk_iter = pd.read_csv(file_name, chunksize=100000)
    
    for i, chunk in enumerate(chunk_iter):
        # Pehle chunk par nayi table banegi (replace), baaki chunks usme add honge (append)
        mode = "replace" if i == 0 else "append"
        
        chunk.to_sql(table_name, engine, if_exists=mode, index=False)
        print(f"  - Chunk {i+1} loaded successfully.")
        
    print(f"✅ '{file_name}' ka poora data load ho gaya!")

print("\n🎉 Badhai ho! Saari 6 tables database mein successfully upload ho gayi hain.")
```
