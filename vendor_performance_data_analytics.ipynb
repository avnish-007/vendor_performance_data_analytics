```python
import pandas as pd
import os
from sqlalchemy import create_engine, inspect
```

```python
# Database Connection
username = "postgres"
password = "your_password" # Apna pgAdmin ka password dalein
host = "localhost"
port = "5432"
database = "vendor_performance_data"

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")
```

```python
# Saari tables ke naam nikaalna aur print karna
inspector = inspect(engine)
tables = inspector.get_table_names()

for table in tables:
    print(table)
```
